```python
import json
with open('./util/result_clean.json','r') as f:
    raw_data = json.load(f)
```

```python
raw_data
```

```python
token_dict = {}
for data in raw_data:
    _list = data['token_sequence']
    for token in _list:
        if token not in token_dict:
            token_dict[token] = 1
        else:
            token_dict[token] += 1
```

```python
sorted_items = sorted(token_dict.items(), key=lambda x: x[1], reverse=True)
sorted_items
```

```python
# 30 이하인 아이템만 필터링
count_le_30 = sum(1 for v in token_dict.values() if v <= 2)

print(f"총 개수가 10 이하인 아이템의 개수: {count_le_30}")
```

```python
len(sorted_items)
```

```python
import matplotlib.pyplot as plt

# 상위 20개만 추출
top_n = 30
top_items = sorted_items[:top_n]
labels, values = zip(*top_items)

# 그래프 그리기
plt.figure(figsize=(12, 6))
plt.barh(labels[::-1], values[::-1])  # 역순으로 그려서 높은 값이 위로 가게
plt.xlabel("Count")
plt.title(f"Top {top_n} Items by Frequency")
plt.tight_layout()
plt.show()
```
